# **1-0**

zip で２つの文字列の同順をペアにし、それぞれ polize, taxi として取り出して c に追加している。これをペアが作れているところまで繰り返す。

```python
a = "パトカー"
b = "タクシー"
d = "猫てゃん"
c = ""
for police, taxi, cat in zip(a, b, d):
  c += police + taxi + cat

print(c)
```

# **1-1**

enumerate で文字列からインデックスと要素を取り出す。そして2, 4, 6, 8文字目を取り出したいので、奇数番目のインデックスを条件に各要素を取り出した。

```python
a = "パタトクカシーー"
b = ""
for i, letter in enumerate(a):
  if i % 2 != 0:
    b += letter

print(b)
```

# **1-4**

dict[キー] = 値　→なければ追加、あれば変更ができる


```python
base_str = "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."
num = [1, 5, 6, 7, 8, 9, 15, 16, 19]
elem_dict = {}
base_list = base_str.split()

for i, item in enumerate(base_list):
    if i + 1 in num:
        elem_dict[item[:1]] = i + 1
    else:
        elem_dict[item[:2]] = i + 1

print(elem_dict)
```
